# Used Car Price Prediction

**Project Objective:**

Develop a **machine learning model to predict the market value of used cars** based on historical data, including vehicle specifications, configurations, and selling prices.

The aim is to achieve the best possible balance between:

* predictive performance;
* prediction speed;
* training time.

**Project Description:**

The used car marketplace **"Not Beaten, Not Painted"** is developing an application that allows users to quickly estimate the market value of their vehicles.

The available dataset contains historical information about used cars, including technical specifications, configurations, and selling prices.

The objective is to build a machine learning model capable of accurately predicting a vehicle's market value.

The client considers the following criteria equally important:

* prediction accuracy;
* prediction speed;
* model training time.

**Dataset Description**:

**Features:**

- **DateCrawled** - date when the advertisement was downloaded from the database.
- **VehicleType** - vehicle body type.
- **RegistrationYear** - year of vehicle registration.
- **Gearbox** - transmission type.
- **Power**- engine power (horsepower).
- **Model** - vehicle model.
- **Kilometer** - mileage (km).
- **RegistrationMonth** - month of vehicle registration.
- **FuelType** - fuel type.
- **Brand** - vehicle manufacturer.
- **Repaired** - whether the vehicle has been repaired.
- **DateCreated** - advertisement creation date.
- **NumberOfPictures** - number of images associated with the advertisement.
- **PostalCode** - postal code of the vehicle owner.
- **LastSeen** - date of the user's last activity.


**Target Variable:**

- **Price** - vehicle price (EUR).

### Additional Analysis of the PostalCode Feature

We initially assumed that the `PostalCode` feature might contain information about the vehicle sale region, which could potentially influence vehicle prices.

To make a final decision about whether to remove this feature, we investigate its relationship with the target variable.

```python
# Calculate Phi-K correlation between PostalCode and Price
correlation = autos_data[['PostalCode', 'Price']].phik_matrix().loc['PostalCode', 'Price']
display(f"Phi-K correlation between PostalCode and Price: {correlation:.3f}")
```

The calculated Phi-K correlation between `PostalCode` and `Price` is **0.096**.

The results do not support our initial hypothesis. Although there is some relationship between postal code (as a possible indicator of region) and vehicle price, the association is relatively weak and explains only a small proportion of the variation.

Therefore, the `PostalCode` feature in its current form is unlikely to provide significant predictive value for the model.

However, before removing this feature completely, we can perform additional analysis. Since postal codes represent geographical areas, we can extract broader regional information by converting postal codes into prefixes corresponding to administrative regions.

Reducing the number of unique categories may decrease noise and allow us to better evaluate the relationship between location and vehicle prices.

Let's explore the unique values of postal codes again:

```python
# list of postal codes
postal_code_list = pd.Series(np.sort(autos_data['PostalCode'].unique()))

display(postal_code_list.head())
display(postal_code_list.tail())
```

**Postal Code Structure Analysis**

The dataset contains 8,143 unique postal codes ranging from 1067 to 99998.

These postal codes correspond to **German postal codes**.

Each administrative region of Germany (federal state) is associated with a range of two-digit postal prefixes.

To identify the approximate region of each advertisement, we create a new feature `plz_prefix` by extracting the first two digits from the original `PostalCode` values.

```python
autos_data_eda = autos_data.copy()


# treat 0 as missing value (if present)
autos_data_eda["PostalCode"] = autos_data_eda["PostalCode"].replace(0, np.nan)

# PLZ prefix (first two digits)
autos_data_eda["plz_prefix"] = (
    autos_data_eda["PostalCode"]
    .dropna()
    .astype(int)
    .astype(str)
    .str.zfill(5)
    .str[:2]
)
```

### Additional Investigation of the `FuelType` Feature

The `FuelType` feature contains both `petrol` and `gasoline`. Since these two terms usually refer to the same fuel in different English-speaking regions, we investigate whether they represent duplicate categories or different fuel types.

```python
autos_data_eda['FuelType'].unique()
```

The fuel type labels are provided in English, although the original dataset appears to originate from Germany.

The **categories** are interpreted as follows:

**Liquid fuels**
- `petrol` - gasoline (British English)
- `gasoline` - gasoline (American English)

**Gaseous fuels**
- `lpg` - Liquefied Petroleum Gas (propane-butane)
- `cng` - Compressed Natural Gas (methane)

**Hybrid vehicles**
- `hybrid` - hybrid vehicles

**Electric vehicles**
- `electric` - battery electric vehicles

**Other fuels**
- `other` - unspecified or uncommon fuel types (e.g., ethanol, hydrogen, or unknown fuels)

Let's examine the **proportions of each fuel type**.

```python
fuel_type_shares = autos_data_eda['FuelType'].value_counts().to_frame()
fuel_type_shares = (fuel_type_shares['count']/fuel_type_shares['count'].sum()).round(4)

fuel_type_shares
```

The distribution of fuel types raises an interesting question.

The dataset distinguishes between **two** gaseous fuel types (`LPG` and `CNG`), despite their relatively small market share, while liquid fuels appear to contain only gasoline-related categories and no separate `diesel` category.

Historically, **diesel** vehicles represented a **substantial share of the German vehicle market**. Around 2016, **diesel** accounted for approximately **45%** of new vehicle registrations and roughly **one-third** of the overall passenger car fleet.

Interestingly, the `gasoline` category represents approximately **30%** of the observations, which is close to the historical market share of **diesel** vehicles.

This suggests that the `gasoline` category may actually correspond to diesel vehicles due to an incorrect translation during dataset preparation.

To evaluate this hypothesis, we compare fuel categories using several independent indicators.

Specifically, we investigate:

- **Fuel distribution across the 20 most common vehicle models.**
  Larger executive cars and commercial vehicles have historically been more likely to use diesel engines, whereas small city cars are predominantly gasoline-powered. If the `gasoline` and `petrol` categories are distributed randomly across vehicle models, this would suggest that they represent the same fuel type. Conversely, if `gasoline` is concentrated among executive and commercial vehicles while `petrol` predominates among compact city cars, this would support the hypothesis that `gasoline` actually represents diesel.

- **Median engine power by fuel type.**
  Diesel vehicles typically have higher engine power than comparable gasoline vehicles. Therefore, if the median engine power of vehicles classified as `gasoline` is substantially higher than that of `petrol`, this would provide additional evidence that `gasoline` corresponds to diesel. Similar median values would instead suggest that both categories represent gasoline.

- **Median and mean mileage by fuel type.**
  Diesel vehicles generally accumulate higher mileage because of their better fuel economy and historically lower operating costs, making them a preferred choice for high-mileage drivers. Consequently, substantially higher median and/or mean mileage for the `gasoline` category would support the diesel hypothesis, whereas similar mileage distributions would suggest that both categories refer to gasoline-powered vehicles.

- **Median vehicle price by fuel type.**
  Diesel vehicles are typically more expensive than comparable gasoline vehicles because they often offer greater engine power and lower fuel consumption. Therefore, if vehicles classified as `gasoline` have consistently higher median prices than those classified as `petrol`, this would provide further evidence that the `gasoline` category actually represents diesel.

```python
# Top 20 most frequent vehicle models
top_models_list = autos_data_eda['Model'].value_counts().head(20).index

test = autos_data_eda[autos_data_eda['Model'].isin(top_models_list)]

# Fuel type distribution across the most common vehicle models
pd.crosstab(
    test['Model'],
    test['FuelType'],
    normalize='index'
).round(2)
```

This distribution **is consistent** with the historical prevalence of diesel engines among larger passenger vehicles and commercial vans in Germany.

**Conclusion**

The dataset does not contain a separate `diesel` category.

However, several independent observations consistently support the hypothesis that the `gasoline` category actually represents diesel-powered vehicles:

- its overall share closely matches the historical share of diesel vehicles in Germany;
- it is considerably more common among larger passenger vehicles and commercial models;
- vehicles classified as `gasoline` have higher median engine power;

Taken together, these findings strongly suggest that `gasoline` is not a duplicate of `petrol`, but rather the result of a possible translation error during dataset preparation.

Therefore, in the following stages of the project, the `gasoline` category will **be renamed** to `diesel`.

```python
def camel_to_snake(name):
    """Convert CamelCase or mixedCase strings to snake_case."""
    name = re.sub('(.)([A-Z][a-z]+)', r'\1_\2', name)
    name = re.sub('([a-z0-9])([A-Z])', r'\1_\2', name)
    return name.lower()
```

##### Imputation Strategy and Order

Before imputing missing values, it is necessary to define both the imputation strategy and the order in which the features will be processed. The first step is to examine the extent of missing data after the previous preprocessing steps. As a result of replacing anomalous values and zeros with NaN, the proportion of missing values increased. The percentage of missing values in each feature is presented below.

Missing values are now present in **7 of the 12 features**, ranging from **4%** to approximately **19%** of all observations.

Because of their prevalence, careful imputation is required. We implement a custom **imputation function** that fills missing values using the **mode**, **mean**, or **median** calculated within groups defined by the most strongly correlated features.

This hierarchical grouping strategy preserves relationships between variables and reduces the bias introduced by global imputation methods, and increases the likelihood that imputed values remain consistent with similar vehicles.

**Rather than relying on global statistics, missing values are imputed hierarchically using the most informative correlated features.**

```python
def fillna_by_group(df, target_col, group_cols, method='mode', fill_label=np.nan):
    """
    Fill missing values in `target_col` using statistics
    computed within groups defined by `group_cols`.

    Parameters
    ----------
    df : pandas.DataFrame
        Input DataFrame.

    target_col : str
        Name of the feature containing missing values.

    group_cols : list of str
        Features used to define the grouping.
        Missing values in these features are temporarily
        replaced with 'unknown_temp' to preserve grouping.

    method : {'mode', 'mean', 'median'}, default='mode'
        Statistic used for imputation.

    fill_label : any, default=np.nan
        Value assigned when no valid observations are
        available within a group.


    Returns
    -------
    pandas.DataFrame
        A new DataFrame with missing values imputed.


    Notes
    -----
    The function performs the following steps:

    1. Creates a copy of the DataFrame.
    2. Temporarily replaces missing values in grouping features
       with 'unknown_temp'.
    3. Groups the data using the specified features.
    4. Computes the selected statistic for each group.
    5. Replaces missing values in `target_col`
       with the corresponding group statistic.
    """

    df = df.copy()
    temp = df[group_cols].apply(lambda col: col.astype(str).fillna('unknown_temp'))
    grouped = df.groupby([temp[c] for c in temp.columns])

    def fill_func(x):
        valid = x.dropna()
        if method == 'mode':
            stat = valid.mode()
            stat_value = stat.iloc[0] if not stat.empty else fill_label
        elif method == 'mean':
            stat_value = valid.mean() if not valid.empty else fill_label
        elif method == 'median':
            stat_value = valid.median() if not valid.empty else fill_label
        else:
            raise ValueError("method must be one of 'mode', 'mean' or 'median'")
        return x.fillna(stat_value)

    df[target_col] = grouped[target_col].transform(fill_func)
    return df
```

##### **Repair History (`repaired`)**

Unlike the previously imputed features, it will not be filled using hierarchical group-based imputation because it exhibits only weak relationships with the remaining predictors. Instead, the feature is treated as an informative categorical variable whose missing values may themselves carry information. This decision is supported by the phik correlation matrix shown below.

```python
repaired_corr = round((phik_corr_4['repaired'].sort_values(ascending=False).to_frame().T), 2)
```

```python
plt.figure(figsize=(19, 1))
sns.heatmap(repaired_corr, annot=True, cmap='coolwarm', center=0)
plt.title("Phik Correlation Matrix for the `repaired` Feature")
plt.show()
```

The feature shows a moderate correlation with the target variable (**price**), suggesting that repaired vehicles generally have lower market prices. However, its relationships with the remaining predictors are weak or negligible.

For this reason, group-based imputation is not appropriate in this case because:

- using the **target variable** would introduce **data leakage**;
- correlations with the remaining predictors are too **weak** to provide reliable estimates;
- missing values may have multiple underlying causes. Some sellers may have intentionally omitted the repair history to conceal previous repairs, whereas others may have assumed that leaving the field blank implicitly indicated that the vehicle had never been repaired (a common behavior in online marketplaces).

Since neither interpretation can be verified, the missing values will be assigned a separate category, 'unknown', rather than being imputed from existing observations.

**Overall Stability of Feature Relationships**

Overall, the pattern of associations between features remained largely stable after missing value imputation.

Most PhiK coefficients changed only slightly (typically by no more than 0.01-0.02), which is an expected outcome after modifying the dataset. Such differences may result from the redistribution of observations within groups, reduced variance after imputation, or the increased completeness of the dataset.

The largest changes were observed for the `repaired` feature.

Compared with the remaining variables, `repaired` exhibited the most noticeable changes:

- associations with `price` and `gearbox` decreased slightly (by approximately 0.01-0.02);
- associations with most other features increased, particularly:
  - `kilometer`: +0.14
  - `model`: +0.09
  - `brand`: +0.05
  - `power`: +0.05

Unlike the other features, missing values in `repaired` were not inferred from correlated observations. Instead, they were assigned to a separate `unknown` category because the missing values could not be interpreted reliably. Consequently, changes in its associations should be interpreted separately from the remaining features.

**Interpretation of the Observed Changes**

The stronger association between `repaired` and `kilometer` is consistent with domain knowledge, since vehicles with higher mileage are generally more likely to experience wear and repairs. Likewise, moderate increases in the associations with `model`, `brand`, and `power` are plausible because these variables are indirectly related to vehicle age, model composition, usage patterns, and maintenance history.

However, these changes should not be interpreted as direct evidence that the imputation improved data quality. Since the missing values in `repaired` were replaced with a separate `unknown` category rather than inferred from neighboring observations, part of the increase in PhiK may also reflect relationships between the lack of repair history and other vehicle characteristics.

**Overall Conclusion**

The comparison of the PhiK association matrices suggests that the selected imputation strategy did not substantially alter the overall association structure of the dataset. Most relationships remained stable, while only a limited number of coefficients changed noticeably.

The largest changes involve the `repaired` feature, whose missing values were handled differently from the remaining variables. Although the observed increases in its associations are plausible from a domain perspective, they should not be interpreted as proof that the imputed dataset is closer to the true underlying data.

Instead, the stability of the overall association structure provides additional evidence that the missing value imputation procedure introduced no major distortions into the dataset. As with any imputation method, changes in correlation coefficients should be interpreted together with domain knowledge and other validation results rather than considered evidence of successful imputation on their own.

#### **LGBM**

`LGBMRegressor` is a regression model based on gradient boosting over decision trees. In boosting, a large number of trees are built sequentially, with each new tree correcting the errors of the previous one.

**Key features of LightGBM**

- **Leaf-wise growth**  
  Trees grow not level by level, but in the direction where the loss reduction is largest. This allows the model to achieve better quality with fewer trees.

- **Histogram-based algorithm**  
  Feature values are converted into 256 bins, which:
  - speeds up training;
  - reduces memory usage;
  - improves model stability.

- **Strong support for categorical features**  
  LightGBM has an internal mechanism for ordering and encoding categories efficiently. Therefore, preprocessing steps such as `OrdinalEncoder`, `OneHotEncoder`, or feature scaling are not required.

**Main hyperparameters of `LGBMRegressor`**

1. `num_leaves` - the number of leaves in each tree. Larger values increase flexibility but also raise the risk of overfitting.
2. `max_depth` - the maximum tree depth, which constrains `num_leaves`.
3. `learning_rate` - the step size of the optimization process. Smaller values usually improve quality but require more trees.
4. `n_estimators` - the number of trees. Smaller `learning_rate` values typically require more estimators.
5. `min_child_samples` - the minimum number of observations per leaf. Increasing this value helps reduce overfitting.
6. `feature_fraction` / `bagging_fraction` - feature and sample subsampling parameters that also help prevent overfitting.
7. `categorical_feature` - the list of categorical columns, which LightGBM handles internally.

For this model, the following **four key hyperparameters** are tuned:

- `num_leaves`
- `max_depth`
- `min_child_samples`
- `n_estimators`

A hyperparameter grid is built, the model is trained, and the key metric, `RMSE`, is calculated using `GridSearchCV`, which automatically evaluates every hyperparameter combination and selects the best score and best parameter set. To improve robustness, cross-validation is used via the `cv` parameter.

`Training time` and `prediction time` are also measured.

**Only four** hyperparameters are used to keep the number of combinations small and reduce training time.

#### Linear Models: Ridge and Lasso

`Linear regression` is a predictive model for numerical targets that seeks a linear relationship between the features and the target variable. It estimates coefficients that minimize the mean squared error (`MSE`).

However, standard linear regression is:

- sensitive to outliers;
- prone to overfitting when the number of features is large;
- unstable in the presence of multicollinearity.

To address these limitations we use the following regularized linear models: **Ridge** and **Lasso**.

**Ridge** uses L2 regularization, which adds a penalty based on the squared coefficient values. This shrinks the model weights without setting them exactly to zero, helping reduce overfitting and improve stability. Ridge works well when the number of features is large.

**Lasso** uses L1 regularization, which adds a penalty based on the absolute values of the coefficients. This can shrink some coefficients to zero, effectively performing feature selection and making the model simpler and more interpretable.

Regularization makes the model more robust, especially when there are many features and strong multicollinearity. For stable model training, numerical features must be scaled using `StandardScaler`, and categorical features must be one-hot encoded.

The main hyperparameter for both models is `alpha`:

- when `alpha` is close to 0, the model behaves similarly to ordinary linear regression;
- when `alpha` is large, the regularization penalty becomes stronger and the weights are reduced more aggressively.

Models are trained for different values of `alpha`, and the best-performing combination is selected using `GridSearchCV`. To improve robustness, cross-validation is used via the `cv` parameter.

| Model | Small `alpha` | Large `alpha` |
|---|---|---|
| Ridge | more flexible | smoother weights |
| Lasso | fewer coefficients set to zero | many coefficients set to zero |

Training time and prediction time are also measured.

Initialize the required tools, start the timer, and train the model.

```python
# Start the timer
start_total_lr = time.time()
```

```python
preprocessor_lr = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), num_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), cat_features)
    ])
```

```python
# Pipeline
pipeline_lr = Pipeline([
    ("preprocess", preprocessor_lr),
    ("model", Ridge())  # base model, it will be replaced by GridSearch.
])
```

```python
# Hyperparameter grid
param_grid = [
    {
        "model": [Ridge(random_state=RANDOM_STATE)],
        "model__alpha": [0.01, 0.1, 1.0, 3.0, 5.0]
    },
    {
        "model": [Lasso(max_iter=10000, random_state=RANDOM_STATE)],
        "model__alpha": [0.01, 0.1, 1.0, 3.0, 5.0]
    }
]
```

```python
# GridSearch
grid_lr = GridSearchCV(
    pipeline_lr,
    param_grid,
    scoring="neg_root_mean_squared_error",
    cv=3,          # small CV to avoid excessive RAM usage
    n_jobs=1,      # set to 1 to avoid excessive RAM usage and prevent a dead kernel
    verbose=1
)
```

#### **Random Forest**

`RandomForestRegressor` is an ensemble model based on building multiple random decision trees, each trained on a different bootstrap sample of the data and features. This reduces overfitting and makes the model more robust to noise and outliers.

The model works well with non-linear relationships and does not require numerical feature scaling. However, categorical data must be encoded using `OneHotEncoder` or `OrdinalEncoder`.

**Key hyperparameters of `RandomForestRegressor`**

1. `n_estimators` - the number of trees in the ensemble. More trees usually improve averaging and stability, but increase training time.
2. `max_depth` - the maximum tree depth. Larger values make the model more flexible but also increase the risk of overfitting. Limiting depth helps stabilize the model.
3. `min_samples_split` - the minimum number of observations required to split an internal node. Larger values reduce overfitting and smooth the model.
4. `min_samples_leaf` - the minimum number of observations required in a leaf node. Increasing this value improves robustness and helps prevent overfitting. This is especially important for noisy data.
5. `max_features` - the number of features considered at each split. Fewer features reduce overfitting and speed up training; more features make individual trees stronger but less diverse.
6. `bootstrap` - whether bootstrap samples are used when training trees. `True` gives the classic random forest setup with high robustness. `False` speeds up training, but trees may be less diverse.
7. `oob_score` - whether out-of-bag samples are used to estimate model quality. This allows evaluation without a separate validation set.
8. `criterion` - the split quality measure:
   - `squared_error` - MSE, the standard option
   - `absolute_error` - more robust to outliers
   - `poisson` - for discrete data
9. `max_leaf_nodes` - the maximum number of leaf nodes in each tree. This limits complexity similarly to `max_depth`.
10. `max_samples` - the size of the bootstrap sample (if `bootstrap=True`). Can be specified as a fraction or an integer. Smaller values reduce training time and overfitting.
11. `ccp_alpha` - the cost-complexity pruning parameter for tree regularization. Larger values produce stronger pruning and reduce overfitting.

Next, we tune three key hyperparameters of the ensemble:

- `n_estimators`
- `max_depth`
- `min_samples_leaf`

For each parameter combination, we train the model and calculate RMSE, then select the configuration with the best performance. `GridSearchCV` is used for this purpose. To improve robustness, cross-validation is applied via the `cv` parameter.

Training time and prediction time are also measured, and the resulting metrics are displayed.

To reduce the number of combinations and shorten training time, **only three** hyperparameters are tuned.

### Comparison of the Best Model (LightGBM) with a Baseline Model

To assess whether the selected LightGBM model provides meaningful predictive value, its test performance is compared with a simple **naive regression baseline**.

A naive regression model does not learn relationships between the features and the target. Instead, it predicts the same constant value for every observation.

Two constant-prediction strategies are examined:

- the **median** target value;
- the **mean** target value.

Because model quality is evaluated using RMSE, the mean is the theoretically optimal constant prediction under squared-error loss. The median is also included for comparison because it is less sensitive to extreme target values.

Let's calculate the RMSE of the naive model using the **median** target value:

```python
# Naive prediction: median price from the training set
naive_pred_value_median = y_train.median()

# Predictions for the test set
y_pred_naive_median = np.full(shape=len(y_test), fill_value=naive_pred_value_median)

# RMSE calculation
#rmse_naive_median = mean_squared_error(y_test, y_pred_naive_median, squared=False)
rmse_naive_median = root_mean_squared_error(y_test, y_pred_naive_median)

display("RMSE of the naive model on the test set (median)", round(rmse_naive_median, 2))
```

Let's calculate the RMSE of the naive model using the **mean** target value:

```python
# Naive prediction: mean price from the training set
naive_pred_value_mean = y_train.mean()

y_pred_naive_mean = np.full(len(y_test), naive_pred_value_mean)

#rmse_naive_mean = mean_squared_error(y_test, y_pred_naive_mean, squared=False)
rmse_naive_mean = root_mean_squared_error(y_test, y_pred_naive_mean)

display("RMSE of the naive model on the test set (mean):", round(rmse_naive_mean, 2))
```

As expected for an RMSE-based comparison, the **mean-value** baseline performs better than the median-value baseline, achieving an RMSE of **€4,505.11** compared with **€4,804.87**.
